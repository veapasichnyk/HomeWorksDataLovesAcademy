#### Завдання 1: Реалізація функції сигмоїди
1. З використанням `numpy` напишіть функцію `sigmoid(z)` для обчислення значення сигмоїди згідно з формулою:
   $$
   \sigma(z) = \frac{1}{1 + e^{-z}}
   $$
2. Використовуючи цю функцію, обчисліть значення сигмоїди для наступних даних: $ z = [-2, -1, 0, 1, 2] $. Виведіть результат обчислень.


```python
import numpy as np
```

```python
#Функція сигмоїди

def sigmoid(z):
    return 1 / (1 + np.exp(-z))
```

```python
z = np.array([-2, -1, 0, 1, 2])
sigmoid_values = sigmoid(z)

print("Sigmoid values:", sigmoid_values)
```



#### Завдання 2: Реалізація функції гіпотези для логістичної регресії
1. Напишіть функцію `hypothesis(theta, X)`, яка обчислює гіпотезу для логістичної регресії, використовуючи функцію сигмоїди. Формула гіпотези:
   $$
   h_\theta(x) = \sigma(\theta^T x) = \frac{1}{1 + e^{-\theta^T x}}
   $$
2. Використайте функцію `hypothesis` для обчислення значень гіпотези для наступних даних:
   
   $\theta = [0.5, -0.5]$
   
   $X = \begin{bmatrix} 1 & 2 \\ 1 & -1 \\ 1 & 0 \\ 1 & 1 \end{bmatrix}$

  Виведіть результат обчислень.


```python
# Функція гіпотези логістичної регресії
def hypothesis(theta, X):
  return sigmoid(np.dot(X, theta ))
```

```python
theta = np.array([0.5, -0.05])
X = np.array([[1, 2], [1, -1], [1, 0], [1, 1]])

h = hypothesis(theta, X)
print("Hypothesis:", h)
```

#### Завдання 3: Реалізація функції для підрахунку градієнтів фукнції втрат
1. Напишіть функцію `compute_gradient(theta, X, y)`, яка обчислює градієнти функції втрат для логістичної регресії. Формула для обчислення градієнта:
   $$
   \frac{\partial L(\theta)}{\partial \theta_j} = \frac{1}{m} \sum_{i=1}^{m} \left[ (h_\theta(x^{(i)}) - y^{(i)}) x_j^{(i)} \right]
   $$
2. Використайте функцію `compute_gradient` для обчислення градієнтів для наступних даних:

  $\theta = [0.5, -0.5]$

  $X = \begin{bmatrix} 1 & 2 \\ 1 & -1 \\ 1 & 0 \\ 1 & 1 \end{bmatrix}$

  $y = [1, 0, 1, 0]$

  Виведіть результат обчислень.

```python
# Функція, яка обчислює градієнти функції втрат для логістичної регресії

def compute_gradient(theta, X, y):
    m = len(y)  # кількість прикладів
    h = hypothesis(theta, X)  # передбачення
    gradient = (1 / m) * np.dot(X.T, (h - y))
    return gradient
```

```python
theta = np.array([0.5, -0.5])
X = np.array([[1, 2], [1, -1], [1, 0], [1, 1]])
y = np.array([1, 0, 1, 0])

grad = compute_gradient(theta, X, y)
print("Gradient:", grad)
```


#### Завдання 4: Реалізація повного батч градієнтного спуску

**Задача:**
1. Напишіть функцію `full_batch_gradient_descent(X, y, lr=0.1, epochs=100)`, яка реалізує алгоритм Full градієнтного спуску для логістичної регресії. Використовуйте такі формули:
   - Гіпотеза: $ h_\theta(x) = \sigma(\theta^T x) $
   - Оновлення параметрів: $ \theta_j := \theta_j - \alpha \frac{\partial L(\theta)}{\partial \theta_j} $
2. Використайте функцію `full_batch_gradient_descent` для обчислення параметрів моделі на наступних даних:

  $X = \begin{bmatrix} 1 & 2 \\ 1 & -1 \\ 1 & 0 \\ 1 & 1 \end{bmatrix}$

  $y = [1, 0, 1, 0]$

  Увага! Матриця $X$ вже має стовпець одиниць і передбачається, що це. - стовпець для intercept - параметра зсуву.

  Виведіть результат обчислень.


```python
#Функція, яка реалізує алгоритм full_batch градієнтного спуску для логістичної регресії

def full_batch_gradient_descent(X, y, lr=0.1, epochs=100):
    n = X.shape[1]                # кількість ознак
    theta = np.zeros(n)           # ініціалізація θ = 0
    loss_history = []

    for _ in range(epochs):
        grad = compute_gradient(theta, X, y)   # ∂L/∂θ
        theta -= lr * grad                     # θ := θ - α * ∂L/∂θ

        # Лог-втрата для перевірки збіжності
        h = hypothesis(theta, X)
        epsilon = 1e-15  # для стабільності логарифма
        loss = -np.mean(y * np.log(h + epsilon) + (1 - y) * np.log(1 - h + epsilon))
        loss_history.append(loss)

    return theta, loss_history
```

```python
X = np.array([[1, 2], [1, -1], [1, 0], [1, 1]])
y = np.array([1, 0, 1, 0])

theta, loss_history = full_batch_gradient_descent(X, y)

print("Навчені параметри θ:", theta)
print("Остання втрата:", loss_history[-1])
```

#### Завдання 5. Обчислення точності моделі

1. Напишіть функцію `predict_proba(theta, X)`, яка використовує знайдені параметри $\theta$ для обчислення ймовірностей належності поточного прикладу з даних до класу $y=1$ на основі значень $\sigma(\theta^T x)$.

2. Напишіть функцію `predict(theta, X, threshold=0.5)`, яка обчислює клас з передбаченої імовірності належності екземпляра до класу 1 з порогом 0.5. Тобто якщо ймовірність менше 0.5, то передбачаємо клас 0, інакше клас 1.

3. Напишіть функцію `accuracy(y_true, y_pred)`, яка обчислює точність моделі, визначивши частку правильно передбачених класів.

  Формула метрики Accuracy:
  $$
  \text{Accuracy} = \frac{\sum_{i=1}^{m} I(\hat{{y}^{(i)}} = y^{(i)})}{m}
  $$

  де $\hat{{y}^{(i)}}$ - передбачене значення класу, $I$ - індикаторна функція (яка дорівнює 1, якщо умова виконується, і 0 - якщо ні), $m$ - кількість прикладів.

4. Обчисліть з використанням даних в завданні 4 $X$, $y$ та обчислених коефіцієнтах $\theta$ та виведіть на екран:
  - передбачені моделлю імовірності належності кожного з екземплярів в матриці `X` до класу 1
  - класи кожного екземпляра з матриці `X`
  - точність моделі.

```python
# Імовірність належності кожного екземпляра до класу 1

def predict_proba(theta, X):
    return sigmoid(np.dot(X, theta))
```

```python
# Обчислюємо клас з передбаченої імовірності належності екземпляра до класу 1 з порогом 0.5

def predict(theta, X, threshold=0.5):
    probs = predict_proba(theta, X)
    return (probs >= threshold).astype(int)
```

```python
#Обраховуємо точність моделі

def accuracy(y_true, y_pred):
    # Перевіряємо, чи однакова довжина вхідних списків
    if len(y_true) != len(y_pred):
        raise ValueError("y_true та y_pred повинні мати однакову довжину.")

    # Обчислюємо кількість правильно передбачених класів
    correct_predictions = sum([1 for true, pred in zip(y_true, y_pred) if true == pred])

    # Повертаємо точність як частку правильно передбачених класів
    return correct_predictions / len(y_true)
```

```python
# Обчислюємо ймовірності та передбачаємо класи
probabilities = predict_proba(theta, X)
predictions = predict(theta, X)
acc = accuracy(y, predictions)

# Виведення результатів
print("Імовірності належності кожного екземпляра до класу 1:")
for i, prob in enumerate(probabilities):
    print(f"Екземпляр {i + 1}: {prob:.4f}")

print("\nПередбачені класи:")
for i, pred in enumerate(predictions):
    print(f"Екземпляр {i + 1}: клас {pred}")

print(f"\nТочність моделі: {acc:.2f}")
```
